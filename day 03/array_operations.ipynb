# Advance Array Multiplication

```python
A=np.array([1,2,3,4]) #you cannot stack matrix
B=np.array([6,7,8,9])
```

```python
np.vstack((A,B))  #vertically stacked
```

```python
np.hstack((A,B))
```

```python
np.column_stack((A,B))
```
