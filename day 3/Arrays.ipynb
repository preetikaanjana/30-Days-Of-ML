# Array Generation Function

```python
arr=np.arange(1,11)
arr
```

```python
arr=np.arange(1,11,2)
arr
```

```python
arr=np.zeros(6)
arr
```

```python
arr=np.zeros((4,8))
arr
```

```python
arr=np.ones(6)
arr
```

```python
arr=np.linspace(1,5,3)  # np.linspace(start, stop, num): Generates 'num' evenly spaced numbers in a range
arr
```

# Random Generation Function

```python
np.random.rand(5) #creates random 5 elements and all value are in 0 and 1 range
```

```python
np.random.randn(10) # values are in range -3 to 3
```

```python
np.random.randint(6) #it will give any random int between 0 to 6
```

```python
np.random.randint(10,20,10)
```
