```python
import numpy as np
```

# Numpy Indexing And Slicing Of Vectors

```python
arr=np.arange(11,21)
arr
```

```python
arr[6]  # Indexing
```

```python
arr[1:4]   # it exectutes till last-1 and if you leave starting value it defaults start from first position and for end value it defaults go till last position
```

```python
arr[2:]
```

```python
arr[3::2]  #skips 2 values in between
```

# Numpy Indexing And Slicing Of Matrix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0] # indexing targets rows 0 prints 0 indexing rows
```

```python
arr[0,0] #for single element
```

```python
arr[0:2,1:3] #slicing
```

```python
arr[3:6,3:5] # or arr[3:,3:]
```

```python
arr[:,2] # for slicing columns
```

# Boolean Indexing

```python
arr=np.arange(11,21)
arr
```

```python
bool_index=arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```
