```python
import numpy as np 
import pandas as pd 
```

# Finding Missing Data

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}
df = pd.DataFrame(data)
```

```python
df
```

```python
df.isna()
```

```python
df.isna().sum()
```

```python
df.isna().any() # Missing Value
```

# Removing Missing Data

```python
# works on the basis or rows
```

```python
df
```

```python
df.dropna()   # all rows with null values will be dropped
```

```python
df
```

```python
df.dropna(thresh=3) # atleast three null value should be presenet in rows
```

# Filling the missing Data

```python
df
```

```python
df.fillna(0) # for applying you need to write inplace = true
```

```python
values = {'A':0,'B':100,"C":300,'D':400}
df.fillna(value=values)
```

```python
# selection : coulumn wise
# fill : coulumn wise
# drop : row wise
```

```python
df.fillna(df.mean())
```
